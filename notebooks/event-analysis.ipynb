```python
import pandas as pd
```

```python
stockprice = pd.read_csv('dataset/stockprice_abnormal_labeled.csv')
news = pd.read_csv('dataset/articles_w_published_dates.csv')
```

```python
stockprice
```

```python
news
```

```python
news['date_published'] = pd.to_datetime(news['date_published'], utc=True).dt.tz_convert('Asia/Jakarta').dt.date
stockprice['Date'] = pd.to_datetime(stockprice['Date']).dt.date
```

```python
df = pd.merge(stockprice, 
              news.drop(columns=['date_accessed', 'link', 'title']), 
              left_on='Date', right_on='date_published', how='inner')
df.drop('date_published', axis=1, inplace=True)
df.drop_duplicates(subset=['content'], inplace=True)
```

```python
df
```

```python
df.to_csv('dataset/sea_labeled.csv', index=False)
```
